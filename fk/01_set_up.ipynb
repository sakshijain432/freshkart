```python
%sql
CREATE CATALOG IF NOT EXISTS fk;
USE CATALOG fk;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS landing;
COMMENT ON SCHEMA landing IS 'Schema for loading data from differnt sources';
CREATE SCHEMA IF NOT EXISTS bronze;
COMMENT ON SCHEMA bronze IS 'Schema for raw data in delta table format';
CREATE SCHEMA IF NOT EXISTS silver;
COMMENT ON SCHEMA silver IS 'Schema for cleaned data in delta table format';
CREATE SCHEMA IF NOT EXISTS gold;
COMMENT ON SCHEMA gold IS 'Schema for dimensional and fact table';
CREATE SCHEMA IF NOT EXISTS ops;
COMMENT ON SCHEMA ops IS 'Schema for operational tables';
```
