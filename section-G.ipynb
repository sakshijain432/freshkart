```python
#COLUMN MASK -G3
spark.sql(f"""
CREATE FUNCTION mask_email(email STRING)
RETURNS STRING
RETURN CASE WHEN email IS NULL THEN NULL
WHEN email LIKE '%@%' THEN CONCAT(SUBSTRING(email, 1, INSTR(email, '@') - 1), '@***')
ELSE email END         
""")

spark.sql(f"""
CREATE FUNCTION mask_phone(phone STRING)
RETURNS STRING
RETURN CASE WHEN phone IS NULL THEN NULL
WHEN phone LIKE '%-%' THEN CONCAT(SUBSTRING(phone, 1, INSTR(phone, '-') - 1), '-***')
ELSE phone END        
""")

spark.sql(f"""
ALTER TABLE {catalog}.gold.dim_customers
ALTER COLUMN email SET MASKED WITH (FUNCTION = mask_email(email))
ALTER TABLE {catalog}.gold.dim_customers
ALTER COLUMN phone SET MASKED WITH (FUNCTION = mask_phone(phone))           
          
""")

```
